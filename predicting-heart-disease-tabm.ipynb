```python
train = pd.read_csv('/kaggle/input/playground-series-s6e2/train.csv',index_col='id')
train.head()
```

```python
test = pd.read_csv('/kaggle/input/playground-series-s6e2/test.csv')
test.head()
```

```python
test_ids = test['id']
```

```python
test.drop('id',axis=1, inplace=True)
```

```python
target = train['Heart Disease'] 
```

```python
train.drop('Heart Disease',inplace=True,axis=1)
```

```python
train.head()
```

```python
target = target.map({'Absence':0,'Presence':1})
```

```python
X = train.copy()
y = target.copy()
```

```python
# from xgboost import XGBClassifier 
from sklearn.metrics import roc_auc_score 
```

```python
TabM_params = {
          'device': 'cuda',
          'batch_size': 'auto',
          'patience': 16,
          'allow_amp': True,
          'arch_type': 'tabm-mini',
          'tabm_k': 32,
          'gradient_clipping_norm': 1.0, 
          'share_training_batches': False,
          'lr': 0.0029993695720154537,
          'weight_decay': 0.023742083301699905,
          'n_blocks': 3,
          'd_block': 448, 
          'dropout': 0.0, 
          'num_emb_type': 'pwl',
          'd_embedding': 32,
          'num_emb_n_bins': 119,
         }
```

```python
from pytabkit import TabM_D_Regressor
from sklearn.metrics import root_mean_squared_error
```

```python
from sklearn.model_selection import StratifiedKFold
N_Splits = 5 
skf = StratifiedKFold(n_splits=5,shuffle=True,random_state=42)

oof_preds = np.zeros(len(X))
test_preds = np.zeros(len(test)) 

for train_idx,val_idx in skf.split(X,y): 
    X_train,X_val = X.iloc[train_idx],X.iloc[val_idx]
    y_train,y_val = y.iloc[train_idx],y.iloc[val_idx]

    model = TabM_D_Regressor(**TabM_params)
    model.fit(X_train, y_train, X_val, y_val)

    # model = XGBClassifier(**xgb_params)
    # model.fit(X_train,y_train) 

    val_preds = model.predict(X_val)
    oof_preds[val_idx] = val_preds 
    
    fold_auc = roc_auc_score(y_val, val_preds)
    print(f"Fold AUC: {fold_auc:.5f}") 
    
    test_preds+= model.predict(test)/N_Splits

cv_auc = roc_auc_score(y, oof_preds)
print(f"\nOverall CV AUC: {cv_auc:.5f}") 

pd.DataFrame({'oof_preds_tabM':oof_preds}).to_csv('oof_preds_tabM.csv',index=False)
pd.DataFrame({'id':test_ids,'test_preds':test_preds}).to_csv('test_preds.csv',index=False)
```
