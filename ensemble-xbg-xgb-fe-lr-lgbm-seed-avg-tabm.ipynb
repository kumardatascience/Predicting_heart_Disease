```python

from scipy.optimize import minimize


# 1. Prepare your data (Use your column names from the screenshot)
models_to_ensemble = ['pred_lgbm_seed', 'pred_tabM', 'pred_xgb', 'pred_lr']
X = X[models_to_ensemble].values  # Your predictions matrix
y                         # Your actual labels (0 or 1)

# 2. Define the 'Objective Function' 
# This is what the computer tries to improve.
def auc_loss(weights):
    # Normalize weights so they always sum to 1.0
    weights = weights / np.sum(weights)
    # Calculate the weighted average
    weighted_pred = np.dot(X, weights)
    # We return negative AUC because the optimizer likes to 'minimize' things
    return -roc_auc_score(y, weighted_pred)

# 3. The Optimization "Magic"
# Start with equal weights (0.25 each)
initial_weights = [1.0 / len(models_to_ensemble)] * len(models_to_ensemble)

# Set boundaries (weights must be between 0 and 1)
bounds = [(0, 1)] * len(models_to_ensemble)

# Run the optimizer
result = minimize(auc_loss, initial_weights, method='SLSQP', bounds=bounds)

# 4. Extract and See the Results
best_weights = result.x / np.sum(result.x)
best_auc = -result.fun

print("--- OPTIMIZATION COMPLETE ---")
for i, model in enumerate(models_to_ensemble):
    print(f"{model} weight: {best_weights[i]:.4f}")

print(f"\nYour Best Individual AUC was: pred_lgbm_seed = 0.955228")
print(f"Your NEW Ensemble AUC is: {best_auc:.6f}")
```

```python
# 1. Gather your TEST predictions into a matrix
# (Make sure the order of models is exactly the same as before!)
test_preds_matrix = test_df[models_to_ensemble].values

# 2. Apply the weights (The Recipe)
# np.dot is the fastest way to multiply the matrix by your weights
final_test_predictions = np.dot(test_preds_matrix, best_weights)

# 3. Add them to your test dataframe for safekeeping
test_df['final_ensemble_output'] = final_test_predictions
```
