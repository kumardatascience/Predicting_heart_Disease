```python
train = pd.read_csv('/kaggle/input/playground-series-s6e2/train.csv',index_col='id')
train.head()
```

```python
target = train['Heart Disease'] 
```

```python
train.drop('Heart Disease',inplace=True,axis=1)
```

```python
train.head()
```

```python
target = target.map({'Absence':0,'Presence':1})
```

```python
X = train.copy()
y = target.copy()
```

```python
from lightgbm import LGBMClassifier 
from sklearn.metrics import roc_auc_score 
```

```python

lgbm_params = {
    'objective': 'binary',     # binary classification
    'n_estimators': 5000,       # number of trees
    'learning_rate': 0.02,     # how fast model learns
    'num_leaves': 31,          # model complexity
    'subsample': 0.8,          # row sampling
    'colsample_bytree': 0.8,   # feature sampling
    # 'random_state': 42,
    'n_jobs': -1
}
```

```python
from sklearn.model_selection import StratifiedKFold
N_Splits = 5   

seeds = [42, 2024, 999]

final_oof = np.zeros(len(X))
final_test = np.zeros(len(test))

for seed in seeds:
    
    print(f"\nRunning seed {seed}") 
    lgbm_params['random_state'] = seed 
    oof_preds = np.zeros(len(X))
    test_preds = np.zeros(len(test))
    
    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=seed)
    
    for train_idx, val_idx in skf.split(X, y):
        
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]
        
        model = LGBMClassifier(**lgbm_params)
        model.fit(X_train, y_train)
        
        val_preds = model.predict_proba(X_val)[:, 1]
        oof_preds[val_idx] = val_preds
        
        test_preds += model.predict_proba(test)[:, 1] / 5
    
    seed_auc = roc_auc_score(y, oof_preds)
    print(f"Seed {seed} CV AUC: {seed_auc:.5f}")
    
    final_oof += oof_preds
    final_test += test_preds

# IMPORTANT: divide AFTER loop
final_oof /= len(seeds)
final_test /= len(seeds)

final_auc = roc_auc_score(y, final_oof)
print(f"\nFinal Averaged CV AUC: {final_auc:.5f}")  

pd.DataFrame({'oof_preds_lgbm': final_oof}).to_csv('oof_preds_lgbm.csv', index=False) 
pd.DataFrame({'id': test_ids, 'target': final_test}).to_csv('submission.csv', index=False)
```
